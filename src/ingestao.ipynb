```python
import requests
import json
import datetime
```

```python
url = "http://compras.dados.gov.br/licitacoes/v1/orgaos.json"

resp = requests.get(url)
data = resp.json()

data_save = data["_embedded"]["Orgaos"]
now = datetime.datetime.now().strftime("%Y-%m-%d_%H:%M:%S")

path = f"/Volumes/raw/compras_governamentais/compras_governamentais_raw/orgaos_list/{now}.json"

with open(path, 'w') as open_file:
    json.dump(data_save, open_file)
```
